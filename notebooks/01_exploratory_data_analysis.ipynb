# Phase 3: Exploratory Data Analysis & Statistical Verification
## Predictive Analytics for Early Type-2 Diabetes Risk Detection

This notebook conducts a rigorous statistical examination of the **UCI Early Stage Diabetes Risk Dataset** (Sylhet Diabetes Hospital Outpatient Study).

### Objectives:
1. **Data Hygiene & Schema Conformance**: Null audits, type checking, boundary values.
2. **Target & Demographic Stratifications**: Age distributions, gender balances, target class proportions.
3. **Statistical Hypothesis Testing**:
   - Parametric (Independent $t$-test) & Non-Parametric (Mann-Whitney $U$) comparison of Age.
   - Pearson's $\chi^2$ Tests of Independence for all 15 discrete clinical symptom variables.
   - Haldane-Anscombe corrected **Odds Ratios (OR)** with 95% Wald Confidence Intervals.
   - **Cramér's $V$** effect sizes and inter-feature association matrix.
4. **Clinical Findings & Modeling Directives**: Feature prioritization for Phase 4 preprocessing & Phase 5 model benchmark.

```python
import sys
from pathlib import Path

# Ensure project root is in path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

from src.eda import (
    load_and_validate_data,
    run_categorical_hypothesis_tests,
    run_continuous_tests,
    compute_interfeature_cramers_matrix,
    RAW_DATA_PATH
)

sns.set_theme(style="whitegrid", font_scale=1.05)
```

### 1. Ingestion & Data Hygiene

```python
df = load_and_validate_data(RAW_DATA_PATH)
print(f"Dataset Shape: {df.shape[0]} patients, {df.shape[1]} columns")
print("\nMissing values count:")
print(df.isnull().sum())
df.head()
```

### 2. Target Class Distribution

```python
target_counts = df["class"].value_counts()
print(target_counts)
print("\nPercentages:")
print((target_counts / len(df)) * 100)
```

### 3. Continuous Variable Analysis: Age Differences

```python
cont_results = run_continuous_tests(df)
for k, v in cont_results.items():
    print(f"{k}: {v}")
```

### 4. Categorical Hypothesis Testing (Chi2, Cramér's V, Odds Ratios)

```python
cat_df = run_categorical_hypothesis_tests(df)
cat_df
```

### 5. Pairwise Cramér's V Association Matrix

```python
cramers_matrix = compute_interfeature_cramers_matrix(df)

plt.figure(figsize=(10, 8), dpi=150)
mask = np.triu(np.ones_like(cramers_matrix, dtype=bool), k=1)
sns.heatmap(cramers_matrix, mask=mask, cmap="Blues", annot=True, fmt=".2f", annot_kws={"size": 8})
plt.title("Cramér's V Inter-Feature Association Heatmap", fontsize=13, fontweight="bold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
